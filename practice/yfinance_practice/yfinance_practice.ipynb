```python
q = yf.EquityQuery(
    "and", [
        yf.EquityQuery("eq", ["region", "us"]),
        yf.EquityQuery("gt", ["percentchange", 3])
    ]
)

results = yf.screen(q, sortField="percentchange", sortAsc=False)

results
```

```python
for item in results["quotes"]:
    print(item["symbol"], item.get("displayName"), item.get("regularMarketChangePercent"))
```

```python
ws = yf.WebSocket()
ws.subscribe(["APPL", "GOOG", "NVDA", "MSFT"])
ws.listen(print)
```

```python
import asyncio

async def main():
    ws = yf.AsyncWebSocket(verbose=False)

    await ws.subscribe(["APPL", "GOOG", "NVDA", "MSFT"])

    async def on_message(msg):
        print(msg)

    await ws.listen(on_message)

asyncio.run(main())

# You can also use proxies by specifying yf.config.network.proxy = {"http": "proxy"}
```
